#### 5*. Рассмотрим полином $p(x)=2x^6-51x^4+48x^3+312x^2-576x-100$. Найдите и выведите отдельно ($x$ и $p(x)$):<br>- локальные минимумы,<br>- [глобальный минимум](https://wiki.loginom.ru/articles/global-minimum.html),<br>- локальные максимумы,<br>- глобальный максимум. 

```python

p = 2*x**6 - 51*x**4 + 48*x**3 + 312*x**2 - 576*x - 100

p1 = sym.diff(p, x)  # первая производная
p2 = sym.diff(p1, x)  # вторая производная

critical_points = sym.solve(p1, x)

local_minima = []
local_maxima = []

for x0 in critical_points:
    # Пропускаем комплексные числа
    if not x0.is_real:
        continue

    second_deriv = p2.subs(x, x0)
    p_val = p.subs(x, x0)

    if second_deriv > 0:
        local_minima.append((float(x0), float(p_val)))
    elif second_deriv < 0:
        local_maxima.append((float(x0), float(p_val)))



# Глобальный минимум
global_min = None
if local_minima:
    global_min = local_minima[0]  # предположим, первый — минимальный
    for точка in local_minima:
        if точка[1] < global_min[1]:  # сравниваем по значению p(x)
            global_min = точка

# Глобальный максимум
global_max = None
if local_maxima:
    global_max = local_maxima[0]  # предположим, первый — максимальный
    for точка in local_maxima:
        if точка[1] > global_max[1]:  # сравниваем по значению p(x)
            global_max = точка

# Вывод результатов
print("Локальные минимумы (x, p(x))")
for x0, p_val in local_minima:
    print(f"x = {x0:6.3f}, p(x) = {p_val:9.3f}")

print("\nЛокальные максимумы (x, p(x))")
for x0, p_val in local_maxima:
    print(f"x = {x0:6.3f}, p(x) = {p_val:9.3f}")

print("\nГлобальные экстремумы")
if global_min:
    print(f"Глобальный минимум: x = {global_min[0]:.3f}, p(x) = {global_min[1]:.3f}")
if global_max:
    print(f"Глобальный максимум: x = {global_max[0]:.3f}, p(x) = {global_max[1]:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def p(x):
    return 2*x**6 - 51*x**4 + 48*x**3 + 312*x**2 - 576*x - 100

x = np.linspace(-4, 4, 5000)  # от -4 до 4, 5000 точек
y = p(x)

local_minima = []  # список для хранения (x, y)
local_maxima = []

for i in range(1, len(x) - 1):
    if y[i-1] > y[i] and y[i] < y[i+1]:
        local_minima.append((x[i], y[i]))
    if y[i-1] < y[i] and y[i] > y[i+1]:
        local_maxima.append((x[i], y[i]))

if local_minima:
    global_min = local_minima[0]  # пусть первый будет минимумом
    for точка in local_minima:
        if точка[1] < global_min[1]:  # сравниваем по значению p(x)
            global_min = точка
else:
    global_min = None

if local_maxima:
    global_max = local_maxima[0]  # пусть первый будет максимумом
    for точка in local_maxima:
        if точка[1] > global_max[1]:  # сравниваем по значению p(x)
            global_max = точка
else:
    global_max = None

print("Локальные минимумы (x, p(x))")
for xm, ym in local_minima:
    print(f"x = {xm:6.3f}, p(x) = {ym:9.3f}")

print("Локальные максимумы (x, p(x))")
for xM, yM in local_maxima:
    print(f"x = {xM:6.3f}, p(x) = {yM:9.3f}")

print("Глобальные экстремумы")
if global_min:
    print(f"Глобальный минимум: x = {global_min[0]:.3f}, p(x) = {global_min[1]:.3f}")
if global_max:
    print(f"Глобальный максимум: x = {global_max[0]:.3f}, p(x) = {global_max[1]:.3f}")

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='p(x)', color='deeppink', linewidth=2)

if local_minima:
    x_min, y_min = zip(*local_minima)
    plt.scatter(x_min, y_min, color='g', label='Лок. минимумы', s=50)

if local_maxima:
    x_max, y_max = zip(*local_maxima)
    plt.scatter(x_max, y_max, color='b', label='Лок. максимумы', s=50)

plt.title('График полинома и его экстремумы')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
