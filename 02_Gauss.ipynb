#### 3. Объедините ранее написанный код в функцию `Gauss2x2` для решения системы $2\times 2$. (★☆☆)

```python
def Gauss2x2(A):
    A = A.copy().astype(float)
    c = -A[1, 0] / A[0, 0]
    A[1] = c * A[0] + A[1]
    x2 = A[1, 2] / A[1, 1]
    x1 = (A[0, 2] - A[0, 1] * x2) / A[0, 0]
    return x1, x2
    
x, y = Gauss2x2(A)
print(f"x = {x}, y = {y}")
```

#### 5*. С использованием ранее написанной функции `Gauss2x2`, для матрицы коэффициентов $$\mathbf{A}=\begin{pmatrix} 0.0003 & 3.0\\ 1.0 & 1.0\end{pmatrix}$$ найдите обратную матрицу $\mathbf{A^{-1}}$. Покажите, что найденная матрица действительно является обратной матрице $\mathbf{A}$. (★☆☆)

```python
def Gauss2x2(A):
    A = A.copy().astype(float)
    c = -A[1, 0] / A[0, 0]
    A[1] = c * A[0] + A[1]
    x2 = A[1, 2] / A[1, 1]
    x1 = (A[0, 2] - A[0, 1] * x2) / A[0, 0]
    return x1, x2

A = np.array([[0.0003, 3.0], [1.0, 1.0]])

I = np.eye(2)

inv_col1 = Gauss2x2(np.column_stack([A, I[:, 0]]))
inv_col2 = Gauss2x2(np.column_stack([A, I[:, 1]]))

A_inv = np.column_stack([inv_col1, inv_col2])

print ("Обратная матрица A^-1:")
print (A_inv)

print ("Проверим что полученная матрица является обратной")
print (A@A_inv)
print ("Главня диагональ единичная, всё остальное равно нулю => получили единичную матрицу при умножении")
```

#### 6. Рассмотрим решение системы линейных уравнений для нахождения коэффициентов уравнения химической реакции на примере реакции горения октана, $\require{mhchem} \ce{C8H18}$. Нам нужно найти неизвестные коэффициенты  $x_1$, $x_2$, $x_3$, и $x_4$ в следующем уравнении: $$x_1\ce{C8H18} + x_2\ce{O2 ->} x_3\ce{CO2} + x_4\ce{H2O}$$ Для этих неизвестных можно записать 3 уравнения из условий материального баланса по $\ce{C}$, $\ce{H}$ и $\ce{O}$ соответственно. Например, 1-е уравнение: $$8x_1 - x_3 = 0$$ (Остальные уравнения составьте самостоятельно.) Эта система недоопределена (3 уравнения и 4 неизвестных). Однако, известно, что коэффициенты уравнения реакции определены с точностью до постоянного множителя, поэтому, полученную систему можно дополнить, например, 4-м уравнением $x_1=1$. Теперь система может быть решена. Для этой системы задайте матрицу коэффициентов $\mathbf{A}$ и вектор свободных членов $\mathbf{b}$. Решите полученную систему, используя функцию `solve()` из модуля `numpy.linalg`.

```python
# C: 8x1 - x3 = 0
# H: 18x1 - 2x4 = 0
# O: 2x2 - 2x3 - x4 = 0
# x1 = 1

A = np.array([
    [8, 0, -1, 0],
    [18, 0, 0, -2],
    [0, 2, -2, -1],
    [1, 0, 0, 0]
])

b = np.array([0, 0, 0, 1])

x = np.linalg.solve(A, b)
print(x)
```

#### Найдите натуральное число, умножив на которое вектор коэффициентов $\mathbf{x}$, получится вектор, содержащий только целые числа (с точностью до ошибок округления). Выведите полученное целочисленное решение. (★☆☆)

```python
A = np.array([[8, 0, -1, 0], [18, 0, 0, -2], [0, 2, -2, -1], [1, 0, 0, 0]])
b = np.array([0, 0, 0, 1])
x = np.linalg.solve(A, b)

def find_multiplier(x):
    for m in range(1, 1000):
        if np.allclose(x * m, np.round(x * m)):
            return m
    return None

m = find_multiplier(x)
integer_solution = (x * m).astype(int)
print(integer_solution)
```
