#### <p style='text-align: justify;'>3. Напишите программу, которая находит и печатает все числа, которые могут быть записаны в виде суммы пятых степеней их цифр. Например: $4151 = 4^5 + 1^5 + 5^5 + 1^5$.

```python
def sum_of_fifth_powers(n):
    return sum(int(digit) ** 5 for digit in str(n))
def find_numbers():
    max_limits = 6*9**5
    result = []
    for num in range(1, max_limits):
        if num == sum_of_fifth_powers(num):
            result.append(num)
    return result
numbers = find_numbers()
print (numbers)
```

####  <p style='text-align: justify;'>4. Напишите программу, которая находит и печатает самое большое число, которое является суммой факториалов своих цифр. Например: $145 = 1! + 4! + 5!$. При помощи магии измерьте время работы вашей программы и подумайте, как можно (с правильным диапазоном цикла) уменьшить его до $30\:\mathrm с$ и меньше.

```python
%%time
from math import factorial

def sum_of_factorial(n):
    return (factorial(int(digit)) for digit in str(n))
            
def find_largest_factorial_sum():
    max_limits = 7*factorial(9)
    largest_number = 0
    for num in range(10, max_limits):
        if num == sum_of_factorial(num):
            largest_number = max(largest_number, num)
    return largest_number
                                 
largest = find_largest_factorial_sum()
print(largest)
```
