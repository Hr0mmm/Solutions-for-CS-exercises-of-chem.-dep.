#### 1. Экспериментальным путём определите машинное эпсилон (★☆☆)

```python
eps = 1

while (1+eps>1):
    eps /= 2
print (eps*2)
```

#### 2. Решите (с использованием дискриминанта) квадратное уравнение $ax^2 + bx + c = 0$ для случая $a = 1.0$, $b = 1.00\cdot10^8$, $c = 1.0$. Сравните результат с точным ответом ($x_1= -1.0000 \cdot 10^{-8}, x_2 = -1.0000\cdot10^{8}$). Объясните полученную разницу.(★☆☆)

```python
a = 1.0
b = 1.00*10**8
c = 1.0
D = b**2 - (4*a*c)
D2 = D**0.5
x1 = (-b+D2)/(2*a)
x2 = (-b-D2)/(2*a)
print (x1, x2)
```

#### * Перепишите программу, чтобы суммирование велось до максимально возможной точности (до тех пор, пока сумма не перестанет изменяться). Выведите количество слагаемых и сравните результат с предыдущим случаем.

```python
import math

for x in [6.0, -6.0, -18.0]:
    s = 1.0  # сумма
    a = 1.0  # член ряда
    
    for k in range(1, 51):
        a = a * x / k
        s = s + a
    
    exact = math.exp(x)
    print(f"x = {x:5.1f}: вычислено = {s:.10e}, точно = {exact:.10e}, ошибка = {s-exact:.10e}")

print ("*")

for x in [6.0, -6.0, -18.0]:
    s_old = 0.0
    s = 1.0
    a = 1.0
    k = 1
    
    while s != s_old:  
        s_old = s
        a = a * x / k
        s = s + a
        k = k + 1
    
    exact = math.exp(x)
    print(f"x = {x:5.1f}: членов = {k-1}, вычислено = {s:.10e}, точно = {exact:.10e}, ошибка = {s-exact:.10e}")
```

#### 4.  Напишите программу для вычисления первых 20 членов ряда $E_n = 1 - n \cdot E_{n-1}$, где $E_1 = e^{-1}$. Оцените разумность полученных значений. (★☆☆)

```python
E = [0.0] * 21  
E[1] = math.exp(-1)  

print(f"n=1: {E[1]:.15f}")

for n in range(2, 21):
    E[n] = 1 - n * E[n-1]
    print(f"n={n}: {E[n]:.15f}")
```

#### Напишите программу для вычисления первых 20 членов того же ряда, воспользовавшись обратной формулой: $E_{n-1} = \frac{1-E_n}{n}$, начав с оценки $E_{30} \approx 0.0$. Оцените разумность полученных значений. (★☆☆)

```python

E = [0.0] * 31  
E[30] = 0.0

for n in range(30, 1, -1):
    E[n-1] = (1 - E[n]) / n


for n in range(1, 21):
    print(f"n={n}: {E[n]:.15f}")
```

#### 5*.  Напишите программу для вычисления значения функции $f(x) = x \left(\sqrt{x+1}-\sqrt{x}\right)$. Оцените, возможны ли ошибки округления при вычислении значения этой функции в какой-либо области значений аргумента. Перепишите функцию, чтобы избежать ошибок округления и сравните результаты работы старой и новой функций для одинакового значения $x$ из проблемной области. (★☆☆)

```python
def f(x):
    return x * (math.sqrt(x+1) - math.sqrt(x))

def f1(x):
    return x / (math.sqrt(x+1) + math.sqrt(x))

print(f(5e15), f1(5e15))
```
